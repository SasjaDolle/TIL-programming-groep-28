# 1. Introduction and motivation

In the Netherlands many railway lines have a motorway running close alongside them, which
makes rail and road alternatives for largely the same journeys. That alternative becomes
relevant whenever a line is out of service for days at a time, which happens in two ways. An
unplanned closure follows from a defect or an incident and is known a day or two in advance at
most; a planned closure is scheduled for engineering work, announced months beforehand and
served by replacement buses (Rijden de Treinen, n.d.). Neither is rare: in 2025, ten unplanned
disruptions lasted three days or more and 52 lasted a day or longer (own calculation from
Rijden de Treinen, 2026). For as long as a closure lasts there is no train on the affected
section, and its passengers have to travel in some other way.

The passengers of a closed line have four alternatives available: the car, the replacement
bus, travel at a different hour, or no travel at all. Only the first of these is observable
from outside the railway system, because the motorways running alongside many Dutch railway
lines are equipped with measurement locations that record traffic intensity per hour
(Nationaal Dataportaal Wegverkeer, n.d.). A substantial shift towards the car should therefore
be detectable as additional traffic on those motorways during the hours in which trains would
normally have operated.

The magnitude of that shift is relevant in two respects. For infrastructure managers it
determines whether a maintenance window can be scheduled independently or requires
coordination with the road authority and avoidance of the commuting peak, and it quantifies
the road capacity on which the rail network implicitly depends whenever a line is closed. In
the research literature the question remains open for closures of this type. Existing
estimates concern urban public transport and are based almost exclusively on strikes: average
highway delay increased by 47 percent during a transit strike in Los Angeles in 2003, with the
largest effects on freeways parallel to the busiest transit lines (Anderson, 2014); 71 transit
strikes in the five largest German cities between 2002 and 2011 have been analysed comparably
(Bauernschuster et al., 2017); and citywide strikes in Rotterdam increased travel times
slightly on the ring road and more on inner-city roads, most strongly during peak hours, while
raising cycling volumes by approximately as much as car volumes (Adler &amp; van Ommeren, 2016).
The present case differs in three respects. It concerns a regional railway line with a
parallel motorway rather than urban public transport. The closures last three to eight days
rather than hours, which is long enough for travellers to adopt a different routine. And
because a strike is announced only one or two days in advance, none of these studies compares
a closure known months beforehand with one announced the preceding evening.

Three Dutch corridors were selected in which a railway line runs alongside a motorway and for
which both a planned and an unplanned multi-day closure could be identified on the same
section of road. For each of the six closures, the hourly traffic intensity measured on the
motorway during the week of the closure is compared with the corresponding hours of an average
week, together with a line indicating the capacity of the road. Section 2 states the research
question and its four components, section 4 the data, section 5 the selection of corridors and
closures, and section 6 the method and its assumptions. Section 7 presents the eight figures
with a reading of each, sections 8 and 9 provide the answers, and section 10 discusses what
could still invalidate them.

# 2. Research objective and questions

The objective of this report is to establish whether a complete multi-day closure of a railway
line is measurable as additional traffic on the motorway running beside it, and whether it
matters how much warning travellers had.

**Main question**

> To what extent do planned and unplanned railway closures affect the traffic intensity on the
> parallel motorway, and does that additional traffic bring the road near its capacity?

**Sub-questions**

1. **How does the traffic intensity during a closure week compare with that of an average
   week on the same corridor?** This is the descriptive core, and it also establishes whether
   the road comes near the capacity of its lanes, because on a road that is already full extra
   demand cannot show up as extra vehicles. Answered in sections 7.2 to 7.4 and 8.1.
2. **Does an announced closure differ from an unannounced one?** A planned closure is known
   months in advance and comes with replacement buses; an unplanned one is known a day or two
   ahead and does not. This is the behavioural question, and it is the one the existing
   literature on transit strikes cannot answer. Answered in section 8.2.
3. **In which hours of the day does any difference appear?** A shift from rail to road should
   sit in the hours the trains would have run, so the shape of the day matters as much as its
   total. Answered in section 8.3.
4. **Do the three corridors differ, and which characteristics of a route could explain that?**
   Number of lanes, how busy the road already is and the size of the railway line beside it
   are the candidates. Answered in section 8.4.

Each sub-question is answered from the same comparison, described in section 6, and from the
same eight figures in section 7. No sub-question is introduced here that the report does not
answer, and no conclusion in section 9 appears that is not one of these four.

# 3. Scope

**Geographical scope.** The study covers three corridors in which a railway line and a
motorway run alongside each other: Utrecht to 's-Hertogenbosch beside the A2, Dordrecht to
Roosendaal beside the A17, and Maastricht to Heerlen beside the A79. Each corridor is
represented not by the whole length of its motorway but by a single cross-section of it, the
one at which the traffic measurement locations used here are situated: hectometre 94.5 and
95.1 on the A2, 7.5 and 7.8 on the A17, and 1.4 and 1.8 on the A79 (Nationaal Dataportaal
Wegverkeer, n.d.). Where both carriageways are measured, the two directions are added together,
so the figures report the total traffic passing that point rather than the traffic in the
direction of the closed railway line.

**Temporal scope.** Six closures are examined, one planned and one unplanned per corridor,
between April 2022 and October 2025. For each closure two whole weeks are used, running Monday
to Sunday: the week in which the closure fell and an average reference week for the same
corridor. The grain of the measurement is one hour, which is the finest the export provides,
so a shift that happens within an hour cannot be seen.

**Unit of analysis.** One hour on one corridor, expressed as the number of vehicles passing
the measurement locations in that hour, summed over the locations available there. Each hour is
compared with the same hour of the week in the reference week, and with the capacity of the
road, taken as the number of lanes multiplied by 2,200 vehicles per hour.

**Outside the scope.** Four things are deliberately not addressed. The passengers who responded
to a closure in some other way than by car, by not travelling, by travelling at another hour or
by cycling, cannot be seen in traffic counts at all, so the measured effect is a lower bound on
the behavioural response. The replacement buses use the same motorway and are counted as
vehicles like any other, which is a mechanism rather than an error but means the rise cannot be
attributed to cars alone. Speed and travel time are not examined, only intensity, so congestion
is addressed only through the distance to the capacity line. And no statistical test is applied,
so the report describes differences but does not establish their significance.

**The disruption archive** (Rijden de Treinen, 2026) is the starting point. Every reported
disruption appears once, with the cause in Dutch and English, a cause group, and the duration
in minutes. It records that a disruption was reported and how long it lasted, but not how much
of the service was still running during it, which is why the check in section 5 needs a
different source. International connections are listed in the same file and are removed here,
because a disruption between Maastricht and Liege says nothing about a Dutch motorway.

**The planned closures** come from the archive of engineering works (Rijden de Treinen, n.d.),
which gives the section, the start and the end of each announced closure but no cause, since
the cause is the work itself.

**The traffic data** are Dexter exports (Nationaal Dataportaal Wegverkeer, n.d.), one ZIP per
weekday per corridor per closure. Each ZIP holds an Excel workbook with four sheets, of which
two are used: `Overzicht`, which names the period, the weekdays and the measurement locations,
and `Intensiteit`, which gives the number of vehicles per hour of the day. The workbooks also
contain average speed and a data quality sheet; neither is used here, which is why speed falls
outside the scope.

Two properties of this export shape the whole method. The first is that an export is always an
**average per hour of the day over a requested period**, not a single measurement. For a
closure week that period is a single date, so the figure is that day. For a reference week the
period is a window of two to five weeks before the closure, filtered to one weekday, so the
reference Tuesday is the average of two to five Tuesdays rather than of one. The reference is
therefore more stable than a single week would be, but the windows differ in length between
the six cases, so the references are not equally precise.

The second is that each corridor is represented by one or two measurement locations rather than
a stretch of road, as section 3 sets out. Where two locations are present they are the two
carriageways of the same cross-section and their counts are added together.

# 6. Method

> **TO WRITE** &middot; about 350 words.

Write out the chain in prose, no code:

1. Each Dexter export is a ZIP with an Excel workbook; the date comes from the `Overzicht`
   sheet and the hourly intensities from the `Intensiteit` sheet.
2. The rows that are hourly intervals are kept and the starting hour is taken from them.
3. The days of the closure week and of the average reference week are laid on one
   Monday-to-Sunday timeline, so each hour can be compared to its counterpart.
4. Measurements taken at the same moment by more than one signal are summed, which is why
   the intensity is a corridor total and not a single loop.
5. The capacity line in each figure is the number of lanes times 2200 vehicles per hour.

the workbook layout is as expected, the average week is a fair counterfactual, 2200 vehicles

commuting is mostly one way.

**Limitations.** Each one with the direction of the problem: does it make the effect look too
big or too small? Naming a limitation honestly costs nothing and hiding one costs the
technical quality mark. At least these six:

1. **One week against one reference week.** Every closure week carries its own weather, its own

2. **The holiday overlap.** All three planned closures sit in or next to a school holiday:
   2 to 9 January 2023, 28 April to 4 May 2022 and 7 to 10 July 2023. That is not a
   coincidence, because major engineering work is scheduled in holidays, but it means the
   planned closures are compared against ordinary weeks that did contain commuting. This
   probably pushes the planned effect down and may explain the whole planned against unplanned
   difference. Check the exact dates against the school holiday calendar and cite it.
3. **No statistical test**, so nothing can be called significant.
4. **Whether the trains really stopped**, from section 5.4.
5. **A loop counts vehicles, not travellers**, and cannot tell a diverted passenger from a lorry.
6. **The capacity line covers both carriageways** while commuting is mostly one direction, so
   one direction can be at capacity while the total looks comfortable. On a road near capacity
   extra demand does not show up as extra vehicles at all, which means a flat result there
   cannot be read as "no shift".
