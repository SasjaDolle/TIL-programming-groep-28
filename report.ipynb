# 1. Introduction and motivation

In 2025 the Dutch rail network recorded 6,676 disruptions. Almost all of them are short: the
median one lasted 49 minutes, and a traveller who misses one train takes the next. The
interesting cases are the exceptions. After international connections are set aside, 52
disruptions in 2025 lasted a day or longer and 10 lasted three days or more, seven of those
caused by problems with the rolling stock and three by repair works. On such days the line is
not delayed but gone, and the people who normally use it have to do something else: take the
car, take a replacement bus, travel at another hour, or not travel at all. Which of those they
choose decides whether a closure is a problem for the railway alone or also for the road
beside it, and that is what makes it a question for maintenance planning, for road operators
and for the business case of rail investment.

That substitution has been measured before, but almost always in cities and almost always
through strikes. Anderson (2014) used a sudden strike by Los Angeles transit workers in 2003
and found that average highway delay rose by 47 percent, with the largest effects on freeways
running parallel to the busiest transit lines. Bauernschuster et al. (2017) studied 71 public
transit strikes in the five largest German cities between 2002 and 2011. Closest to the Dutch
situation, Adler and van Ommeren (2016) used citywide transit strikes in Rotterdam and found
the effect small on the ring road but larger on inner-city roads and strongest in the rush
hour, and they also found that a strike raises the number of cyclists about as much as the
number of car drivers.

Three things are missing from that work, and together they are what this report is about. The
cases are urban transit, not a regional railway line with a motorway running beside it. The
disruptions are strikes, which by their nature come with a day or two of notice, so none of
that research compares a closure people knew about for months with one they heard about the
evening before. And the disruptions last hours or a day, not the three to eight days that a
complete engineering closure takes, which is long enough for travellers to settle into a new
routine rather than improvise once.

This report therefore takes three Dutch corridors where a railway line runs alongside a
motorway, selects one planned and one unplanned multi-day closure on each, and compares the
hourly traffic intensity measured on that motorway during the week of the closure with the
same hours of an average week.

Section 4 sets out the four data sources and what each of them can and cannot show. Section 5
explains how the three corridors and the six closures were chosen, and whether the trains
really stopped running on those days. Section 6 describes the comparison and the assumptions
it rests on. Section 7 presents the eight figures, two for the national picture and two per
corridor, each with a reading. Sections 8 and 9 answer the sub-questions and then the main
question, and section 10 discusses what could still make that answer wrong.

# 6. Method

> **TO WRITE** &middot; about 350 words.

Write out the chain in prose, no code:

1. Each Dexter export is a ZIP with an Excel workbook; the date comes from the `Overzicht`
   sheet and the hourly intensities from the `Intensiteit` sheet.
2. The rows that are hourly intervals are kept and the starting hour is taken from them.
3. The days of the closure week and of the average reference week are laid on one

5. The capacity line in each figure is the number of lanes times 2200 vehicles per hour.

# 10. Discussion

> **TO WRITE** &middot; about 450 words.

**What this means.** Who could use this and for what decision: maintenance planning, the
business case for rail, road management during works. Two or three sentences, no overclaiming.

**Limitations.** Each one with the direction of the problem: does it make the effect look too
big or too small? Naming a limitation honestly costs nothing and hiding one costs the
technical quality mark. At least these six:

1. **One week against one reference week.** Every closure week carries its own weather, its own
   incidents and its own calendar.
2. **The holiday overlap.** All three planned closures sit in or next to a school holiday:
   2 to 9 January 2023, 28 April to 4 May 2022 and 7 to 10 July 2023. That is not a
   coincidence, because major engineering work is scheduled in holidays, but it means the
   planned closures are compared against ordinary weeks that did contain commuting. This
   probably pushes the planned effect down and may explain the whole planned against unplanned
   difference. Check the exact dates against the school holiday calendar and cite it.
3. **No statistical test**, so nothing can be called significant.
4. **Whether the trains really stopped**, from section 5.4.
5. **A loop counts vehicles, not travellers**, and cannot tell a diverted passenger from a lorry.
6. **The capacity line covers both carriageways** while commuting is mostly one direction, so
   one direction can be at capacity while the total looks comfortable. On a road near capacity
   extra demand does not show up as extra vehicles at all, which means a flat result there
   cannot be read as "no shift".

**Next steps.** What you would do with another month: more closures per corridor, a test with
a noise floor, speed as well as intensity, and the train archive used per section per day.
Concrete, not "more research is needed".

# Appendix. Where the code is and how to reproduce it

The analysis is in `main.ipynb`. It reads the disruption archive from
`data/disruptions-2025.csv`, filters it, and then loads the traffic exports for the three
corridors. The eight figures in this report were produced by that notebook and saved into
`figures/`, under the names listed in `figures/README.md`.

The traffic data sits in one folder per corridor, `data/data_Utrecht-Den Bosch`,
`data/data_Dordrecht-Roosendaal` and `data/data_Maastricht-Heerlen`. Each holds `Planned`
and `Unplanned`, and each of those holds `Average`, the reference week, and `Delay_week`,
the week of the closure. Inside is one ZIP per weekday, with an Excel workbook in it as
Dexter delivers it. `data/Geplande disruption periodes.xlsx` is the export the planned
closure dates were taken from; no code reads it, it documents where those dates came from.
`data/trajectwerkzaamheden langer dan 1 dag.csv` is written by the notebook, not read by it.

`Archive/` holds files that are no longer part of the analysis, with a README saying per
file why it is there.
