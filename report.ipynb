# 1. Introduction and motivation

**Situation**<br>
In the Netherlands many railway lines have a motorway running close alongside them, which
makes rail and road alternatives for largely the same journeys. That alternative becomes
relevant whenever a line is out of service for days at a time, which happens in two ways. An
unplanned closure follows from a defect or an incident and is known a day or two in advance at
most; a planned closure is scheduled for engineering work, announced months beforehand and
served by replacement buses (Rijden de Treinen, n.d.). Neither is rare: in 2025, 52 domestic disruptions
lasted a day or longer and ten lasted three days or more; of those ten, seven were unplanned
and three were engineering work (own calculation from Rijden de Treinen, 2026). For as long as a closure lasts there is no train on the affected
section, and its passengers have to travel in some other way.

![Duration of disruptions in 2025](figures/01_duration_threshold.png)

<p style="color:#6b7280;font-size:0.88em"><b>Figure 1.</b> Every disruption recorded in 2025, by the date it began and
how long it lasted. The dashed line marks three days. Almost the entire distribution sits far
below it; the handful of points above are the closures this report is about.</p>

**Question**<br>
The passengers of a closed line have four alternatives available: the car, the replacement
bus, travel at a different hour, or no travel at all. Only the first of these is observable
from outside the railway system, because the motorways running alongside many Dutch railway
lines are equipped with measurement locations that record traffic intensity per hour
(Nationaal Dataportaal Wegverkeer, n.d.). A substantial shift towards the car should therefore
be detectable as additional traffic on those motorways during the hours in which trains would
normally have operated.

**Relevance**<br>
The magnitude of that shift matters for infrastructure managers, because it determines whether
a maintenance window can be scheduled independently or has to be coordinated with the road
authority and kept out of the commuting peak. For Dutch closures of this type it is largely
unknown. The Dutch evidence comes from citywide public transport strikes in Rotterdam, which
increased travel times slightly on the ring road and more on inner-city roads, most strongly
during peak hours, and raised cycling volumes by approximately as much as car volumes (Adler
&amp; van Ommeren, 2016). That case differs from the present one in three respects: it concerns
urban public transport rather than a regional railway line with a motorway alongside it; it
concerns a strike of hours rather than a closure of several days, which is long enough for
travellers to adopt a different routine; and because a strike is announced one or two days in
advance at most, it cannot show whether a closure known months beforehand differs from one
announced the preceding evening.

**Definition**<br>
Not every disruption is suitable for this purpose. A delay of an hour, or a thinned timetable
on which intercity services still run, leaves most passengers on the train. This report
therefore restricts itself to complete closures lasting several days, in which no service ran
on the affected section, and uses the word closure for those throughout; disruption is kept for
the wider category from which they were selected.

**Approach**<br>
Three Dutch corridors were selected in which a railway line runs alongside a motorway and for
which both a planned and an unplanned closure of this kind could be identified on the same
section of road. For each of the six closures, the hourly traffic intensity measured on the
motorway during the week of the closure is compared with the corresponding hours of an average
week, together with a line indicating the capacity of the road. That comparison is then read in
four ways: as a difference in level, as a difference between an announced and an unannounced
closure, as a difference between the hours of the day, and as a difference between corridors,
which vary both in the length of the closure and in how often they are disrupted in an ordinary
year.


# 2. Research objective and questions

The objective of this report is to establish whether a complete multi-day closure of a railway
line is measurable as additional traffic on the motorway running beside it, and under which
circumstances that effect is larger or smaller.

**Main question**

> To what extent do railway disruptions affect the traffic intensity and congestion on parallel
> highways?

**Sub-questions**

1. **Does it make a difference whether a disruption is planned or unplanned?** A planned
   closure is announced months in advance and served by replacement buses; an unplanned one is
   known a day or two ahead and is not. Three of the six closures are of each kind, one per
   corridor. Answered in section 8.1.
2. **At which time of day is the impact on highway traffic intensity highest?** A shift from
   rail to road should fall in the hours in which the trains would have run, so the shape of the
   day matters as much as its total. The export gives one value per hour, which is the finest
   grain available. Answered in section 8.2.
3. **Does the duration of a disruption affect highway traffic intensity?** The six closures
   last 3.0, 3.0, 3.2, 5.9, 6.9 and 7.6 days, so they fall into a group of roughly three days
   and a group of six to eight. Comparing those two groups is possible, but duration is not
   varied independently of the corridor or of the type of closure, so the answer can only be
   qualitative. Answered in section 8.3.
4. **Is there a difference between routes that are disrupted often and routes that are
   disrupted less?** In 2025 the three corridors recorded 107, 143 and 96 disruptions
   respectively, but 226, 399 and 1,524 hours of disruption in total, so the two ways of
   counting do not agree on which route is the most affected (own calculation from Rijden de
   Treinen, 2026). Answered in section 8.4.

Each sub-question is answered from the same comparison, set out in section 6, and from the same
six figures in section 7. The conclusion in section 9 answers these four and the main
question, and nothing else.


# 3. Scope

**Geographical scope.** The study covers three corridors in which a railway line and a
motorway run alongside each other: Utrecht to 's-Hertogenbosch beside the A2, Dordrecht to
Roosendaal beside the A17, and Maastricht to Heerlen beside the A79. On the second of these
the stretch that closed is Lage Zwaluwe to Roosendaal, a part of the line rather than all of
it; the corridor is named after its end points throughout. Each corridor is
represented not by the whole length of its motorway but by a single cross-section of it, the
one at which the traffic measurement locations used here are situated: hectometre 94.5 on the A2 for the
planned closure and 95.1 for the unplanned one, 7.5 and 7.8 on the A17, and 1.4 and 1.8 on
the A79 (Nationaal Dataportaal
Wegverkeer, n.d.). Where both carriageways are measured, the two directions are added together,
so the figures report the total traffic passing that point rather than the traffic in the
direction of the closed railway line.

![The three corridors](figures/00_chosen_corridors.jpg)

<p style="color:#6b7280;font-size:0.88em"><b>Figure 2.</b> The three corridors. The railway line is drawn in blue and
the motorway in orange; the solid part of each is the stretch that was closed and the road
alongside it, and an open square marks a station outside the closure. From left to right:
Utrecht Centraal to 's-Hertogenbosch beside the A2, Lage Zwaluwe to Roosendaal beside the A17,
and Maastricht to Heerlen beside the A79.</p>

**Temporal scope.** Six closures are examined, one planned and one unplanned per corridor,
between April 2022 and November 2025, lasting between 3.0 and 7.6 days. For each closure two whole weeks are used, running Monday
to Sunday: the week in which the closure fell and an average reference week for the same
corridor. The grain of the measurement is one hour, which is the finest the export provides,
so a shift that happens within an hour cannot be seen.

**Unit of analysis.** One hour on one corridor, expressed as the number of vehicles passing
the measurement locations in that hour, summed over the locations available there. Each hour is
compared with the same hour of the week in the reference week, and with the capacity of the
road, taken as the number of lanes multiplied by 2,200 vehicles per hour.

**Outside the scope.** Four things are deliberately not addressed. The passengers who responded
to a closure in some other way than by car, by not travelling, by travelling at another hour or
by cycling, cannot be seen in traffic counts at all, so the measured effect is a lower bound on
the behavioural response. The replacement buses use the same motorway and are counted as
vehicles like any other, which is a mechanism rather than an error but means the rise cannot be
attributed to cars alone. Speed and travel time are not examined, only intensity, so congestion
is addressed only through the distance to the capacity line. And no statistical test is applied,
so the report describes differences but does not establish their significance.

**The disruption archive** (Rijden de Treinen, 2026) is the starting point. Every reported
disruption appears once, with the cause in Dutch and English, a cause group, and the duration
in minutes. It records that a disruption was reported and how long it lasted, but not how much
of the service was still running during it, which is why the check in section 5 needs a
different source. International connections are listed in the same file and are removed here,
because a disruption between Maastricht and Liege says nothing about a Dutch motorway.

**The planned closures** come from the archive of engineering works (Rijden de Treinen, n.d.),
which gives the section, the start and the end of each announced closure but no cause, since
the cause is the work itself.

**The traffic data** are Dexter exports (Nationaal Dataportaal Wegverkeer, n.d.), one ZIP per
weekday per corridor per closure. Each ZIP holds an Excel workbook with four sheets, of which
two are used: `Overzicht`, which names the period, the weekdays and the measurement locations,
and `Intensiteit`, which gives the number of vehicles per hour of the day. The workbooks also
contain average speed and a data quality sheet; neither is used here, which is why speed falls
outside the scope.

Two properties of this export shape the whole method. The first is that an export is always an
**average per hour of the day over a requested period**, not a single measurement. For a
closure week that period is a single date, so the figure is that day. For a reference week the
period is a window of two to five weeks before the closure, filtered to one weekday, so the
reference Tuesday is the average of two to five Tuesdays rather than of one. The reference is
therefore more stable than a single week would be, but the windows differ in length between
the six cases, so the references are not equally precise.

The second is that each corridor is represented by one or two measurement locations rather than
a stretch of road, as section 3 sets out. Where two locations are present they are the two
carriageways of the same cross-section and their counts are added together.

# 6. Method

**The comparison.** Each closure is assessed against a reference for the same corridor, hour
by hour. The week of the closure and the reference week are laid on one Monday-to-Sunday
timeline, so that every hour is compared with the same hour of the same weekday, and the two
are plotted over each other. A rise caused by the closure should therefore appear as a gap
between the two lines, concentrated in the hours the trains would have run.

**What one observation is.** A Dexter export is an average per hour of the day over a
requested period, not a single reading. For a closure week each export covers one date, so the
line is that day. For a reference week each export covers a window of two to five weeks before
the closure, filtered to one weekday, so the reference Tuesday is the mean of several Tuesdays.
The reference is therefore steadier than a single week would be. Where a corridor has two
measurement locations, one per carriageway, their counts are added, so each point is the total
traffic passing that cross-section in that hour.

**The capacity line.** Every figure carries a horizontal line at the number of lanes times
2,200 vehicles per hour, the conventional capacity of a motorway lane. It is there because
intensity measures what passes, which is the smaller of what drivers want and what the road can
take; the distance to that line says how much room there was for extra traffic to appear in.

**Assumptions.** Four, each of which returns in section 10. The workbooks are laid out as
expected, with the date in `Overzicht` and hourly values in `Intensiteit`. The weeks before a
closure are a fair counterfactual for the week of it. A lane carries 2,200 vehicles per hour.
And the two carriageways may be added, although commuting runs mainly in one direction at a
time.

**No test is applied.** The comparison is read from the figures. Differences are therefore
described, not established, and the words significant and caused by are avoided throughout.

Across all three routes, the main observation is that traffic intensity during the selected

parallel highway traffic intensity during railway track closures, nor do they establish a clear
difference between planned and unplanned disruptions.

# 10. Discussion

**What this means in practice.** If a closure of several days leaves no visible mark on the
motorway beside it, the road authority need not be involved in planning the maintenance window,
and the spare capacity the railway relies on during a closure is larger than assumed. A visible
mark in the commuting peak would imply the opposite, and would make the timing of engineering
work a joint decision rather than a railway one.

**Limitations.** Five, each with the direction in which it pushes the result.

1. **One closure against one short reference.** Weather, incidents, roadworks and the holiday
   calendar are not controlled for, so each comparison carries the noise of a single week. For
   the unplanned A2 case the reference window also overlaps the closure itself, which pulls
   that difference towards zero.
2. **The planned closures fall in or beside school holidays.** Engineering work is scheduled
   in holidays on purpose, so those closure weeks had less commuting of their own. This
   depresses the measured effect of a planned closure and may account for part of any
   difference found between the two kinds.
3. **Two of the three cases selected as unplanned are recorded as engineering work in the
   source.** The contrast between announced and unannounced closures is therefore weaker than
   the design intended, and sub-question 1 should be read with that in mind.
4. **One cross-section, both directions added.** A corridor is represented by the traffic
   passing one point, with the two carriageways summed, so a shift that runs in one direction
   during the peak is diluted by the quiet direction. On a road already near its capacity,
   extra demand does not appear as extra vehicles at all, which means a flat line is not
   evidence that nobody switched.
5. **A detector counts vehicles, not travellers.** Passengers who stayed home, cycled or
   travelled at another hour are invisible, and the replacement buses use the same road. The
   measured effect is therefore a lower bound on the behavioural response, and part of it is
   bus rather than car.

**Next steps.** Three would change the answer rather than refine it: verify in the timetable
archive that no train ran on the days selected, which this report assumes; measure each
carriageway separately so that a one-directional shift is not averaged away; and extend the
sample beyond six closures, so that a difference can be tested instead of described.

# Appendix. Where the code is and how to reproduce it

The analysis is in `main.ipynb`. It reads the disruption archive from
`data/disruptions-2025.csv`, filters it, and then loads the traffic exports for the three
corridors. Figures 1 and 3 to 9 were produced by that notebook and saved into `figures/`, under the
names listed in `figures/README.md`; Figure 2 comes from the midterm presentation.
`Actual_Services.ipynb` checks, from the realised services archive, whether trains really
stopped running, and `Filtering_Service_data.ipynb` reduces that archive to the per-closure
files in the repository root.

The traffic data sits in one folder per corridor, `data/data_Utrecht-Den Bosch`,
`data/data_Dordrecht-Roosendaal` and `data/data_Maastricht-Heerlen`. Each holds `Planned`
and `Unplanned`, and each of those holds `Average`, the reference week, and `Delay_week`,
the week of the closure. Inside is one ZIP per weekday, with an Excel workbook in it as
Dexter delivers it. `data/Geplande disruption periodes.xlsx` is an earlier exploration of
planned closures in 2024, none of which were used; no code reads it.
`data/trajectwerkzaamheden langer dan 1 dag.csv` is written by the notebook, not read by it.
