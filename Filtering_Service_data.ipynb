```python
import pandas as pd
```

```python
Services_2022 = pd.read_csv('data/Actual_Services_data/services-2022.csv')
Services__January_2023 = pd.read_csv('data/Actual_Services_data/services-2023-01.csv')
Services_July_2023 = pd.read_csv('data/Actual_Services_data/services-2023-07.csv')
Services_June_2025 = pd.read_csv('data/Actual_Services_data/services-2025-06.csv')
Services_October_2025 = pd.read_csv('data/Actual_Services_data/services-2025-10 (1).csv')
Services_November_2025 = pd.read_csv('data/Actual_Services_data/services-2025-11.csv')
```

```python
Line_Numbers_Maastricht_Heerlen = ("32", "189")
Line_Numbers_DenBosch_Utrecht = ("88", "27", "29", "35", "39", "60")
Line_Numbers_Roosendaal_Dordrecht = ("59", "22", "23")
```

```python
Filtered_2022 = Services_2022[Services_2022["Service:Train number"].astype(str).str.startswith(Line_Numbers_DenBosch_Utrecht, na=False)]
Filtered_2022.to_csv("2022-utr-db.csv", index=False)

Filtered_Jan_2023 = Services__January_2023[Services__January_2023["Service:Train number"].astype(str).str.startswith(Line_Numbers_Maastricht_Heerlen, na=False)]
Filtered_Jan_2023.to_csv("jan-2023-mst-hrl.csv", index=False)

Filtered_Jul_2023 = Services_July_2023[Services_July_2023["Service:Train number"].astype(str).str.startswith(Line_Numbers_Roosendaal_Dordrecht, na=False)]
Filtered_Jul_2023.to_csv("jul-2023-dor-roo.csv", index=False)

Filtered_Jun_2025 = Services_June_2025[Services_June_2025["Service:Train number"].astype(str).str.startswith(Line_Numbers_Roosendaal_Dordrecht, na=False)]
Filtered_Jun_2025.to_csv("jun-2025-dor-roo.csv", index=False)

Filtered_Oct_2025 = Services_October_2025[Services_October_2025["Service:Train number"].astype(str).str.startswith(Line_Numbers_Maastricht_Heerlen, na=False)]
Filtered_Oct_2025.to_csv("jun-2025-mst-hrl.csv", index=False)

Filtered_Nov_2025 = Services_November_2025[Services_November_2025["Service:Train number"].astype(str).str.startswith(Line_Numbers_DenBosch_Utrecht, na=False)]
Filtered_Nov_2025.to_csv("nov-2025-utr-db.csv", index=False)
```
